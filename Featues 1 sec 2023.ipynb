```python
import pandas as pd
import os
```

```python
data_1sec_2023 = pd.read_csv('final_combined_2023_with_time.csv')
```

```python
from IPython.display import display, HTML

html_table = data_1sec_2023.head(10).to_html(index=False)
display(HTML(html_table))
```

```python
data_1sec_2023['Mid_Price_1'] = (data_1sec_2023['Ask Price 1'] + data_1sec_2023['Bid Price 1']) / 2
data_1sec_2023['Mid_Price_2'] = (data_1sec_2023['Ask Price 2'] + data_1sec_2023['Bid Price 2']) / 2
data_1sec_2023['Spread_1'] = data_1sec_2023['Ask Price 1'] - data_1sec_2023['Bid Price 1']
data_1sec_2023['Spread_2'] = data_1sec_2023['Ask Price 2'] - data_1sec_2023['Bid Price 2']
data_1sec_2023['Volume_Mean_1'] = (data_1sec_2023['Ask Size 1'] + data_1sec_2023['Bid Size 1']) / 2
data_1sec_2023['Volume_Mean_2'] = (data_1sec_2023['Ask Size 2'] + data_1sec_2023['Bid Size 2']) / 2
data_1sec_2023['Order_Imbalance_1'] = (data_1sec_2023['Bid Size 1'] - data_1sec_2023['Ask Size 1']) / (data_1sec_2023['Bid Size 1'] + data_1sec_2023['Ask Size 1'])
data_1sec_2023['Order_Imbalance_2'] = (data_1sec_2023['Bid Size 2'] - data_1sec_2023['Ask Size 2']) / (data_1sec_2023['Bid Size 2'] + data_1sec_2023['Ask Size 2'])
data_1sec_2023['VWAP_1'] = (data_1sec_2023['Ask Price 1'] * data_1sec_2023['Ask Size 1'] + 
                             data_1sec_2023['Bid Price 1'] * data_1sec_2023['Bid Size 1']) / \
                            (data_1sec_2023['Ask Size 1'] + data_1sec_2023['Bid Size 1'])

data_1sec_2023['VWAP_2'] = (data_1sec_2023['Ask Price 2'] * data_1sec_2023['Ask Size 2'] + 
                             data_1sec_2023['Bid Price 2'] * data_1sec_2023['Bid Size 2']) / \
                            (data_1sec_2023['Ask Size 2'] + data_1sec_2023['Bid Size 2'])
```

```python
data_1sec_2023.to_csv('data_1sec_2023_with_features.csv', index=False)
```
