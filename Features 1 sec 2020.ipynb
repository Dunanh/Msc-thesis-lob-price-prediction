```python
import pandas as pd
import os
```

```python
data_1sec_2020 = pd.read_csv('final_combined_2020_with_time.csv')
```

```python
from IPython.display import display, HTML

html_table = data_1sec_2020.head(10).to_html(index=False)
display(HTML(html_table))
```

```python
# Assuming you have a DataFrame 'data' with 'Direction' as the target column
print("Class distribution in the entire dataset:")
print(data_1sec_2020['Direction'].value_counts())
```

```python
data_1sec_2020['Mid_Price_1'] = (data_1sec_2020['Ask Price 1'] + data_1sec_2020['Bid Price 1']) / 2
data_1sec_2020['Mid_Price_2'] = (data_1sec_2020['Ask Price 2'] + data_1sec_2020['Bid Price 2']) / 2
data_1sec_2020['Spread_1'] = data_1sec_2020['Ask Price 1'] - data_1sec_2020['Bid Price 1']
data_1sec_2020['Spread_2'] = data_1sec_2020['Ask Price 2'] - data_1sec_2020['Bid Price 2']
data_1sec_2020['Volume_Mean_1'] = (data_1sec_2020['Ask Size 1'] + data_1sec_2020['Bid Size 1']) / 2
data_1sec_2020['Volume_Mean_2'] = (data_1sec_2020['Ask Size 2'] + data_1sec_2020['Bid Size 2']) / 2
data_1sec_2020['Order_Imbalance_1'] = (data_1sec_2020['Bid Size 1'] - data_1sec_2020['Ask Size 1']) / (data_1sec_2020['Bid Size 1'] + data_1sec_2020['Ask Size 1'])
data_1sec_2020['Order_Imbalance_2'] = (data_1sec_2020['Bid Size 2'] - data_1sec_2020['Ask Size 2']) / (data_1sec_2020['Bid Size 2'] + data_1sec_2020['Ask Size 2'])
data_1sec_2020['VWAP_1'] = (data_1sec_2020['Ask Price 1'] * data_1sec_2020['Ask Size 1'] + 
                             data_1sec_2020['Bid Price 1'] * data_1sec_2020['Bid Size 1']) / \
                            (data_1sec_2020['Ask Size 1'] + data_1sec_2020['Bid Size 1'])

data_1sec_2020['VWAP_2'] = (data_1sec_2020['Ask Price 2'] * data_1sec_2020['Ask Size 2'] + 
                             data_1sec_2020['Bid Price 2'] * data_1sec_2020['Bid Size 2']) / \
                            (data_1sec_2020['Ask Size 2'] + data_1sec_2020['Bid Size 2'])
```

```python
data_1sec_2020.to_csv('data_1sec_2020_with_features.csv', index=False)
```
