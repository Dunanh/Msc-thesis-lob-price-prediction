```python
import pandas as pd
import os
```

```python
data_1sec_2017 = pd.read_csv('final_combined_2017_with_time.csv')
```

```python
from IPython.display import display, HTML

html_table = data_1sec_2017.head(10).to_html(index=False)
display(HTML(html_table))
```

Mid Price

```python
data_1sec_2017['Mid_Price_1'] = (data_1sec_2017['Ask Price 1'] + data_1sec_2017['Bid Price 1']) / 2
data_1sec_2017['Mid_Price_2'] = (data_1sec_2017['Ask Price 2'] + data_1sec_2017['Bid Price 2']) / 2
```

Spread

```python
data_1sec_2017['Spread_1'] = data_1sec_2017['Ask Price 1'] - data_1sec_2017['Bid Price 1']
data_1sec_2017['Spread_2'] = data_1sec_2017['Ask Price 2'] - data_1sec_2017['Bid Price 2']
```

Volume Means 


```python
data_1sec_2017['Volume_Mean_1'] = (data_1sec_2017['Ask Size 1'] + data_1sec_2017['Bid Size 1']) / 2
data_1sec_2017['Volume_Mean_2'] = (data_1sec_2017['Ask Size 2'] + data_1sec_2017['Bid Size 2']) / 2
```

Order Imbalance

```python
data_1sec_2017['Order_Imbalance_1'] = (data_1sec_2017['Bid Size 1'] - data_1sec_2017['Ask Size 1']) / (data_1sec_2017['Bid Size 1'] + data_1sec_2017['Ask Size 1'])
data_1sec_2017['Order_Imbalance_2'] = (data_1sec_2017['Bid Size 2'] - data_1sec_2017['Ask Size 2']) / (data_1sec_2017['Bid Size 2'] + data_1sec_2017['Ask Size 2'])
```

VWAP

```python
data_1sec_2017['VWAP_1'] = (data_1sec_2017['Ask Price 1'] * data_1sec_2017['Ask Size 1'] + 
                             data_1sec_2017['Bid Price 1'] * data_1sec_2017['Bid Size 1']) / \
                            (data_1sec_2017['Ask Size 1'] + data_1sec_2017['Bid Size 1'])

data_1sec_2017['VWAP_2'] = (data_1sec_2017['Ask Price 2'] * data_1sec_2017['Ask Size 2'] + 
                             data_1sec_2017['Bid Price 2'] * data_1sec_2017['Bid Size 2']) / \
                            (data_1sec_2017['Ask Size 2'] + data_1sec_2017['Bid Size 2'])
```

```python
data_1sec_2017.to_csv('data_1sec_2017_with_features.csv', index=False)
```
