```python
resampled5min2016 = pd.read_csv('resampled_5min_final_2016_corrected.csv')

#print(resampled5min2016.head())
```

##### Mid-price

```python
# Level 1
resampled5min2016['Mid_Price_1_first'] = (resampled5min2016['Ask Price 1_first'] + resampled5min2016['Bid Price 1_first']) / 2
resampled5min2016['Mid_Price_1_last'] = (resampled5min2016['Ask Price 1_last'] + resampled5min2016['Bid Price 1_last']) / 2
resampled5min2016['Mid_Price_1_min'] = (resampled5min2016['Ask Price 1_min'] + resampled5min2016['Bid Price 1_min']) / 2
resampled5min2016['Mid_Price_1_max'] = (resampled5min2016['Ask Price 1_max'] + resampled5min2016['Bid Price 1_max']) / 2

# Level 2
resampled5min2016['Mid_Price_2_first'] = (resampled5min2016['Ask Price 2_first'] + resampled5min2016['Bid Price 2_first']) / 2
resampled5min2016['Mid_Price_2_last'] = (resampled5min2016['Ask Price 2_last'] + resampled5min2016['Bid Price 2_last']) / 2
resampled5min2016['Mid_Price_2_min'] = (resampled5min2016['Ask Price 2_min'] + resampled5min2016['Bid Price 2_min']) / 2
resampled5min2016['Mid_Price_2_max'] = (resampled5min2016['Ask Price 2_max'] + resampled5min2016['Bid Price 2_max']) / 2

```

##### Spread

```python
# Level 1
resampled5min2016['Spread_1_first'] = resampled5min2016['Ask Price 1_first'] - resampled5min2016['Bid Price 1_first']
resampled5min2016['Spread_1_last'] = resampled5min2016['Ask Price 1_last'] - resampled5min2016['Bid Price 1_last']
resampled5min2016['Spread_1_min'] = resampled5min2016['Ask Price 1_min'] - resampled5min2016['Bid Price 1_min']
resampled5min2016['Spread_1_max'] = resampled5min2016['Ask Price 1_max'] - resampled5min2016['Bid Price 1_max']

# Level 2
resampled5min2016['Spread_2_first'] = resampled5min2016['Ask Price 2_first'] - resampled5min2016['Bid Price 2_first']
resampled5min2016['Spread_2_last'] = resampled5min2016['Ask Price 2_last'] - resampled5min2016['Bid Price 2_last']
resampled5min2016['Spread_2_min'] = resampled5min2016['Ask Price 2_min'] - resampled5min2016['Bid Price 2_min']
resampled5min2016['Spread_2_max'] = resampled5min2016['Ask Price 2_max'] - resampled5min2016['Bid Price 2_max']


```

##### Price Means

```python
# Level 1
resampled5min2016['Price_Mean_1_first'] = (resampled5min2016['Ask Price 1_first'] + resampled5min2016['Bid Price 1_first']) / 2
resampled5min2016['Price_Mean_1_last'] = (resampled5min2016['Ask Price 1_last'] + resampled5min2016['Bid Price 1_last']) / 2
resampled5min2016['Price_Mean_1_min'] = (resampled5min2016['Ask Price 1_min'] + resampled5min2016['Bid Price 1_min']) / 2
resampled5min2016['Price_Mean_1_max'] = (resampled5min2016['Ask Price 1_max'] + resampled5min2016['Bid Price 1_max']) / 2

# Level 2
resampled5min2016['Price_Mean_2_first'] = (resampled5min2016['Ask Price 2_first'] + resampled5min2016['Bid Price 2_first']) / 2
resampled5min2016['Price_Mean_2_last'] = (resampled5min2016['Ask Price 2_last'] + resampled5min2016['Bid Price 2_last']) / 2
resampled5min2016['Price_Mean_2_min'] = (resampled5min2016['Ask Price 2_min'] + resampled5min2016['Bid Price 2_min']) / 2
resampled5min2016['Price_Mean_2_max'] = (resampled5min2016['Ask Price 2_max'] + resampled5min2016['Bid Price 2_max']) / 2
```

##### Volume Means

```python
# Level 1
resampled5min2016['Volume_Mean_1'] = (resampled5min2016['Ask Size 1_sum'] + resampled5min2016['Bid Size 1_sum']) / 2

# Level 2
resampled5min2016['Volume_Mean_2'] = (resampled5min2016['Ask Size 2_sum'] + resampled5min2016['Bid Size 2_sum']) / 2
```

##### Order Imbalance

```python
# Level 1
resampled5min2016['Order_Imbalance_1'] = (resampled5min2016['Bid Size 1_sum'] - resampled5min2016['Ask Size 1_sum']) / (resampled5min2016['Bid Size 1_sum'] + resampled5min2016['Ask Size 1_sum'])

# Level 2
resampled5min2016['Order_Imbalance_2'] = (resampled5min2016['Bid Size 2_sum'] - resampled5min2016['Ask Size 2_sum']) / (resampled5min2016['Bid Size 2_sum'] + resampled5min2016['Ask Size 2_sum'])
```

##### Volume Weighted Avereage Price (VWAP)

```python
# Level 1
resampled5min2016['VWAP_1'] = (resampled5min2016['Ask Price 1_last'] * resampled5min2016['Ask Size 1_sum'] + 
                               resampled5min2016['Bid Price 1_last'] * resampled5min2016['Bid Size 1_sum']) / \
                              (resampled5min2016['Ask Size 1_sum'] + resampled5min2016['Bid Size 1_sum'])

# Level 2
resampled5min2016['VWAP_2'] = (resampled5min2016['Ask Price 2_last'] * resampled5min2016['Ask Size 2_sum'] + 
                               resampled5min2016['Bid Price 2_last'] * resampled5min2016['Bid Size 2_sum']) / \
                              (resampled5min2016['Ask Size 2_sum'] + resampled5min2016['Bid Size 2_sum'])
```
