```python
import pandas as pd
import os
data_1sec_2016 = pd.read_csv('final_combined_2016_with_time.csv')
data_1sec_2016 = data_1sec_2016.drop(columns=['Date'])
```

```python
from IPython.display import display, HTML

html_table = data_1sec_2016.head(10).to_html(index=False)
display(HTML(html_table))
```

##### Mid-Price

```python
data_1sec_2016['Mid_Price_1'] = (data_1sec_2016['Ask Price 1'] + data_1sec_2016['Bid Price 1']) / 2
data_1sec_2016['Mid_Price_2'] = (data_1sec_2016['Ask Price 2'] + data_1sec_2016['Bid Price 2']) / 2
```

##### Spread

```python
data_1sec_2016['Spread_1'] = data_1sec_2016['Ask Price 1'] - data_1sec_2016['Bid Price 1']
data_1sec_2016['Spread_2'] = data_1sec_2016['Ask Price 2'] - data_1sec_2016['Bid Price 2']
```

##### Volume Means

```python
data_1sec_2016['Volume_Mean_1'] = (data_1sec_2016['Ask Size 1'] + data_1sec_2016['Bid Size 1']) / 2
data_1sec_2016['Volume_Mean_2'] = (data_1sec_2016['Ask Size 2'] + data_1sec_2016['Bid Size 2']) / 2
```

##### Order Imbalance 

```python
data_1sec_2016['Order_Imbalance_1'] = (data_1sec_2016['Bid Size 1'] - data_1sec_2016['Ask Size 1']) / (data_1sec_2016['Bid Size 1'] + data_1sec_2016['Ask Size 1'])
data_1sec_2016['Order_Imbalance_2'] = (data_1sec_2016['Bid Size 2'] - data_1sec_2016['Ask Size 2']) / (data_1sec_2016['Bid Size 2'] + data_1sec_2016['Ask Size 2'])
```

##### VWAP

```python
data_1sec_2016['VWAP_1'] = (data_1sec_2016['Ask Price 1'] * data_1sec_2016['Ask Size 1'] + 
                             data_1sec_2016['Bid Price 1'] * data_1sec_2016['Bid Size 1']) / \
                            (data_1sec_2016['Ask Size 1'] + data_1sec_2016['Bid Size 1'])

data_1sec_2016['VWAP_2'] = (data_1sec_2016['Ask Price 2'] * data_1sec_2016['Ask Size 2'] + 
                             data_1sec_2016['Bid Price 2'] * data_1sec_2016['Bid Size 2']) / \
                            (data_1sec_2016['Ask Size 2'] + data_1sec_2016['Bid Size 2'])
```

```python
data_1sec_2016.to_csv('data_1sec_2016_with_features.csv', index=False)
```
